# Analyzing 2019-2025 ATL Hawks Data

### Importing Data

```python
import pandas as pd

df = pd.read_csv('/Users/sophiabraskamp/Documents/Student_Game_Data.csv')

print(df.shape)
print(df.head())
print(df.dtypes)
```

### Adding column for repeat customers

```python
df['customer_type'] = df.groupby('acct_id')['acct_id'].transform('count').apply(lambda x: 'repeat' if x > 1 else 'single')
df
```

### Customer Type Breakdown by Opponents

```python
by_opponent = df.groupby('Opponent')['customer_type'].value_counts(normalize=True) * 100
print(by_opponent)
```

### Customer Type Breakdown by Event Name

```python
by_event = df.groupby('event_name')['customer_type'].value_counts(normalize=True) * 100
print(by_event)
```
